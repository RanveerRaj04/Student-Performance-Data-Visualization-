```python
import pandas as pd
df=pd.read_csv("student_performance.csv")
print(df)
```

```python
print("no. of students =",df.shape[0])
print("average study hours =",df["Study_Hours"].sum()/df.shape[0])
print("average sleep hours =",df["Sleep_Hours"].sum()/df.shape[0])
print("average attendance =",df["Attendance"].sum()/df.shape[0])
print("average marks in each subjects:")
print("         Maths :",round(df["Math"].mean(),2))
print("         Physics :",round(df["Physics"].mean(),2))
print("         Chemistry :",round(df["Chemistry"].mean(),2))
print("         English :",round(df["English"].mean(),2))
print("         Computer_Science :",round(df["Computer_Science"].mean(),2))
print("Student with highest computer science marks : Student ID:",
      df[df["Computer_Science"]==df["Computer_Science"].max()]["Student_ID"].iloc[0])
print("Student with lowest math marks : Student ID:",
      df[df["Math"]==df["Math"].min()]["Student_ID"].iloc[0])
```

```python
import matplotlib.pyplot as plt
x=df["Student_ID"].values
y=df["Computer_Science"].values

plt.title("Student ID vs Computer Science marks",fontweight="bold",fontsize=16)
plt.xlabel("Student_ID",fontweight="bold",fontsize=12)
plt.ylabel("Computer Science Marks",fontweight="bold",fontsize=12)
plt.plot(x,y,marker="o",markersize=10,mfc="black")
plt.grid(linestyle="dotted",alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
x=["Math","Physics","Chemistry","English","Computer Science"]
y=[df["Math"].mean(),df["Physics"].mean(),df["Chemistry"].mean(),df["English"].mean(),df["Computer_Science"].mean()]
bars=plt.bar(x,y,color="maroon")
plt.bar_label(bars,padding=2)
plt.title("Average Marks in each Subject",fontsize=15)
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.tight_layout()
plt.grid(axis="y",alpha=0.6,linestyle="dotted")
plt.show()
```

```python
plt.hist(df["Study_Hours"],edgecolor="darkgreen",bins=10)
plt.grid(alpha=0.6,linestyle="dotted")
plt.title("Study Hours",fontsize=16)
plt.xlabel("No. of hours")
plt.ylabel("No. of students")
plt.show()
```

```python
df["Average_Marks"]=(df["Math"]+df["Physics"]+df["Chemistry"]+df["English"]+df["Computer_Science"])/5
x=df["Study_Hours"]
y=df["Average_Marks"]
plt.scatter(x,y)
plt.grid(alpha=0.6,linestyle="dotted")
plt.title("Study_Hours v/s Average_Marks")
plt.xlabel("Study_Hours")
plt.ylabel("Average_Marks")
plt.show()

print("This scatter plot shows that Study hours has a positive correlation with Average marks")
```

```python
df["Average_Marks"]=(df["Math"]+df["Physics"]+df["Chemistry"]+df["English"]+df["Computer_Science"])/5
x=df["Sleep_Hours"]
y=df["Average_Marks"]
plt.scatter(x,y)
plt.grid(alpha=0.6,linestyle="dotted")
plt.title("Sleep_Hours v/s Average_Marks")
plt.xlabel("Sleep_Hours")
plt.ylabel("Average_Marks")
plt.show()

print("This scatter plot shows that Study hours has a positive correlation with Average marks")
```

```python
x=df["Student_ID"].values
y=df["Attendance"].values

plt.title("Student ID vs Attendance",fontsize=16)
plt.xlabel("Student ID",fontsize=12)
plt.ylabel("Attendance",fontsize=12)
plt.xticks(x)
plt.axhline(y=df["Attendance"].mean(),label=df["Attendance"].mean(),color="red")
bars=plt.bar(x,y)
plt.bar_label(bars)
plt.show()
```

```python
print("Highest attendance : Student ID:",df[df["Attendance"]==df["Attendance"].max()]["Student_ID"].iloc[0])
print("Lowest attendance : Student ID:",df[df["Attendance"]==df["Attendance"].min()]["Student_ID"].iloc[0])
print("Average attendance =",df["Attendance"].mean())
```

```python
fig,ax=plt.subplots(2,2)
ax[0,0].hist(df["Study_Hours"].values,edgecolor="darkblue",bins=10)
ax[0,0].set_title("no. of students vs Study Hours")

ax[0,1].scatter(df["Study_Hours"].values,df["Average_Marks"].values,color="red",edgecolor="black")
ax[0,1].set_title("Study_Hours vs Average_Marks")

ax[1,0].scatter(df["Sleep_Hours"].values,df["Average_Marks"].values,color="blue",edgecolor="darkgreen")
ax[1,0].set_title("Sleep_Hours vs Average_Marks")

ax[1,1].bar(df["Student_ID"].values,df["Attendance"].values,color="yellow",edgecolor="gold")
ax[1,1].set_title("Student_ID vs Attendance")

plt.tight_layout()
```
